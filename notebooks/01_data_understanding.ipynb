
# E-Commerce Business Intelligence & Customer Analytics

## Project Overview
This project analyzes the Brazilian E-Commerce Public Dataset by Olist to uncover insights into sales performance, customer behaviour, product performance, payments, reviews and delivery operations.

## Objectives
1. Understand the structure of the datasets.
2. Inspect data quality and missing values.
3. Explore relationships between the datasets.
4. Identify important business metrics.
5. Prepare the data for SQL analytics and Power BI dashboards.

## Dataset Source
Brazilian E-Commerce Public Dataset by Olist — Kaggle.

## Tools
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- SQL
- Power BI


```python

# Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# Configure notebook display

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:,.2f}".format)

# Set visualization style

sns.set_theme(style="whitegrid")

# Define the raw data directory

RAW_DATA = Path("../data/raw")

print("Libraries imported successfully!")
print("Raw data directory:", RAW_DATA.resolve())
print("Directory exists:", RAW_DATA.exists())
```

```python

# Identify all CSV datasets

csv_files = sorted(RAW_DATA.glob("*.csv"))

dataset_inventory = pd.DataFrame({
    "Dataset": [file.name for file in csv_files],
    "Size (MB)": [
        round(file.stat().st_size / (1024 ** 2), 2)
        for file in csv_files
    ]
})

print("TOTAL DATASETS:", len(csv_files))

display(dataset_inventory)
```

```python

# Load the eight core Olist datasets

datasets = {
    "customers": "olist_customers_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "category_translation": "product_category_name_translation.csv"
}

# Read each CSV file into a Pandas DataFrame

data = {}

for name, filename in datasets.items():
    file_path = RAW_DATA / filename
    data[name] = pd.read_csv(file_path)

    print(f"Loaded {name}: {data[name].shape[0]:,} rows")

print("\nAll eight core datasets loaded successfully!")
```

```python

# Examine the dimensions of each dataset

dataset_dimensions = pd.DataFrame([
    {
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    }
    for name, df in data.items()
])

display(dataset_dimensions)
```

```python

# Inspect the columns and data types of all eight datasets

for name, df in data.items():

    print("\n" + "=" * 65)
    print(f"DATASET: {name.upper()}")
    print("=" * 65)

    print("Shape:", df.shape)

    print("\nColumn names and data types:")
    print(df.dtypes)

    print("\nFirst 3 records:")
    display(df.head(3))
```

```python

# Assess data quality across the eight datasets

quality_results = []

for name, df in data.items():

    quality_results.append({
        "Dataset": name,
        "Total Rows": len(df),
        "Missing Values": df.isna().sum().sum(),
        "Columns With Missing Values": df.isna().any().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

quality_summary = pd.DataFrame(quality_results)

display(quality_summary)
```

```python

# Detailed missing value analysis

for name, df in data.items():

    missing = df.isna().sum()
    missing = missing[missing > 0]

    if not missing.empty:

        missing_report = pd.DataFrame({
            "Missing Values": missing,
            "Missing Percentage": (
                missing / len(df) * 100
            ).round(2)
        }).sort_values(
            by="Missing Values",
            ascending=False
        )

        print(f"\nMISSING VALUES: {name.upper()}")
        display(missing_report)

    else:
        print(f"\n{name.upper()}: No missing values")
```

```python
# Investigating missing product information

product_missing = data["products"][
    data["products"]["product_category_name"].isna()
]

print("PRODUCT MISSING VALUE INVESTIGATION")
print("-" * 45)

print("Products missing category:",
      data["products"]["product_category_name"].isna().sum())

print("Products missing name length:",
      data["products"]["product_name_lenght"].isna().sum())

print("Products missing description length:",
      data["products"]["product_description_lenght"].isna().sum())

print("Products missing photos quantity:",
      data["products"]["product_photos_qty"].isna().sum())

print("\nNumber of products missing all four fields:",
      product_missing[
          [
              "product_name_lenght",
              "product_description_lenght",
              "product_photos_qty"
          ]
      ].isna().all(axis=1).sum())

display(product_missing.head(10))
```

```python
# Investigate product category information

products = data["products"]
translations = data["category_translation"]

print("PRODUCT CATEGORY INVESTIGATION")
print("-" * 50)

print(
    "Missing product categories:",
    products["product_category_name"].isna().sum()
)

print(
    "Unique product categories:",
    products["product_category_name"].nunique()
)

print(
    "Available translated categories:",
    translations["product_category_name_english"].nunique()
)

print("\nCATEGORY TRANSLATION PREVIEW")

display(translations.head(10))
```

```python
# Identify product categories without English translations

product_categories = set(
    products["product_category_name"].dropna().unique()
)

translated_categories = set(
    translations["product_category_name"].dropna().unique()
)

untranslated_categories = product_categories - translated_categories

print("CATEGORY TRANSLATION CHECK")
print("-" * 50)

print("Categories in products:", len(product_categories))
print("Categories in translation table:", len(translated_categories))
print("Categories without translations:", len(untranslated_categories))

print("\nUNTRANSLATED CATEGORY NAMES:")

for category in sorted(untranslated_categories):
    print("-", category)

print("\nDUPLICATE TRANSLATION KEYS:",
      translations["product_category_name"].duplicated().sum())
```

```python
# Add translations for the two missing category names

additional_translations = pd.DataFrame({
    "product_category_name": [
        "pc_gamer",
        "portateis_cozinha_e_preparadores_de_alimentos"
    ],
    "product_category_name_english": [
        "gaming_pc",
        "portable_kitchen_food_processors"
    ]
})

translations = pd.concat(
    [translations, additional_translations],
    ignore_index=True
)

print("UPDATED TRANSLATION TABLE")
print("-" * 50)

print("Total translation records:", len(translations))

print(
    "Unique Portuguese categories:",
    translations["product_category_name"].nunique()
)

print(
    "Duplicate translation keys:",
    translations["product_category_name"].duplicated().sum()
)

display(
    translations.tail(5)
)
```

```python
# Merge English category translations into products

products = data["products"]

products = products.merge(
    translations,
    on="product_category_name",
    how="left"
)

# Save the updated products DataFrame
data["products"] = products

print("CATEGORY TRANSLATION RESULTS")
print("-" * 50)

print("Total products:", len(products))

print(
    "Missing Portuguese categories:",
    products["product_category_name"].isna().sum()
)

print(
    "Missing English categories:",
    products["product_category_name_english"].isna().sum()
)

print("\nTRANSLATION PREVIEW:")

display(
    products[
        [
            "product_category_name",
            "product_category_name_english"
        ]
    ].drop_duplicates().head(10)
)
```

```python
# Correct spelling errors in product column names

products = data["products"]

products = products.rename(columns={
    "product_name_lenght": "product_name_length",
    "product_description_lenght": "product_description_length"
})

data["products"] = products

print("PRODUCT COLUMN NAME VALIDATION")
print("-" * 50)

print("product_name_length exists:",
      "product_name_length" in products.columns)

print("product_description_length exists:",
      "product_description_length" in products.columns)

print("\nUpdated product columns:")

print(products.columns.tolist())
```
